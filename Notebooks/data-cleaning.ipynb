## Data Cleaning

```python
import pandas as pd
```

```python
df = pd.read_csv('shopping_behavior_updated.csv')
```

```python
df.head()
```

**Column Types:**
- **Numerical -** `Customer ID`,  `Age`,  `Purchase Amount (USD)`,  `Review Rating`,  `Previous Purchases` 
    
- **Categorical -** Remaining columns are Categorical columns

- **Mixed -** None — all columns shown are clearly numerical or categorical. 
    

```python
df.shape
```

```python
df.columns
```

```python
# fix column names 
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df = df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})
```

```python
# Check
df.head()
```

```python
df.dtypes
```

```python
# check null values
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
# check invalid values
df['age'].between(18, 70).all()
df['purchase_amount'].between(20, 100).all()
df['review_rating'].between(1, 5).all()
df['previous_purchases'].ge(0).all()
```

```python
# check duplicate record
(df['discount_applied'] == df['promo_code_used']).all()
```

#### drop duplicate `promo_code_used` column
 it is not useful column 

```python
 df = df.drop(columns=['promo_code_used'])
```

## Feature engineering

```python
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']

df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
df['age_group'].value_counts()
```

```python
print(df.shape)
print(df.isnull().sum().sum())
print(df.duplicated().sum())
```

#### Save cleaned Data

```python
df.to_csv('customer_behaviour_cleaned.csv', index=False)
```
